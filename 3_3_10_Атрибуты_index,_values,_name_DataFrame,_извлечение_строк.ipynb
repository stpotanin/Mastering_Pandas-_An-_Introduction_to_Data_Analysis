Допустим, у нас есть список списков с данными о пассажирах авиакомпании, которые были на борту рейса из США в Европу `passengers_data`. Для того, чтобы анализировать данные и делать выводы, удобнее использовать объект DataFrame.

Создайте из списка DataFrame с помощью библиотеки pandas и присвойте ему названия столбцов: `'Age'`, `'Gender'`, `'Country'`. В созданном DataFrame каждый пассажир должен иметь свой уникальный индекс - имя (`'Name'`).

Распечатайте получившийся DataFrame.

```python
import pandas as pd

passengers_data = [['John Smith', 32, 'male', 'USA'],
    ['Sarah Johnson', 28, 'female', 'USA'],
    ['Anna Schmidt', 42, 'female', 'Germany'],
    ['Paul Kowalski', 50, 'male', 'Poland'],
    ['Luigi Rossi', 24, 'male', 'Italy'],
    ['Marie Dupont', 31, 'female', 'France'],
    ['Javier Fernandez', 27, 'male', 'Spain'],
    ['Elena Petrova', 35, 'female', 'Russia'],
    ['Lars Johansson', 47, 'male', 'Sweden']]

df = pd.DataFrame(passengers_data, columns=['Name', 'Age', 'Gender', 'Country'])
df = df.set_index('Name')
print(df)
```
