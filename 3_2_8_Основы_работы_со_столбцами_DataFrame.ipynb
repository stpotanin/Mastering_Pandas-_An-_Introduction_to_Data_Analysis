Дан словарь `data`, с данными о продажах которые относятся к компании, которая работает в области продажи электроники в различных регионах. В таблице содержится информация о продажах нескольких продавцов, включая их имена, возраст, пол, а также общую сумму продаж за 2019 и 2020 годы. Также в таблице указывается регион, в котором работает каждый продавец. Требуется создать на его основе DataFrame с колонками, которые соответствуют ключам словаря и данными, приведенными в списках.

Распечатайте DataFrame, в следующей строке распечатайте его первые две строки, затем последнюю строку.
