Допустим, у нас есть компания, которая занимается продажей фруктов. У нас есть DataFrame, в котором хранятся данные о продажах фруктов за последние несколько месяцев. Колонки в этом DataFrame представляют собой различные атрибуты, такие как дата продажи, тип фрукта, количество проданных единиц и т.д.

Недавно компания приняла решение скрыть информацию о количестве проданных товаров. Для этого нужно удалить колонку, которая соответствует этой информации, чтобы она не отображалась в DataFrame.

Выведите исходный, а затем обновленный DataFrame на печать.

```python
import pandas as pd

data = {'Date': ['2022-01-01', '2022-01-02', '2022-01-03', '2022-01-04', '2022-01-05'],
        'Fruit': ['Apple', 'Banana', 'Orange', 'Apple', 'Orange'],
        'Quantity': [10, 15, 20, 12, 18],
        'Price': [1.5, 2.0, 1.8, 1.2, 2.2]}

df = pd.DataFrame(data)
df = df.set_index('Date')
print(df)

df.drop(['Quantity'], axis=1, inplace=True)
print(df)

# Установите 'Date' в качестве индекса!
```
