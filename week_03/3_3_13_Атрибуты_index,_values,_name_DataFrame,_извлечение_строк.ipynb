Предположим, у нас есть список людей, некоторые из которых любят определенный фрукт в виде списка словарей `data`. Нам нужно создать DataFrame из этого списка.

Также у нас есть объект Series `fruits`, в котором на соответствующих позициях стоят любимые фрукты людей из списка `data`.  Вам нужно добавить данные из объекта Series в DataFrame и затем, вывести его на печать.

```python
import pandas as pd

# Объект Series
fruits = pd.Series(['apple', 'banana', 'orange'], name='fruits')
# Список словарей для создания DataFrame
data = [{'name': 'Alice', 'age': 25},
    {'name': 'Bob', 'age': 30},
    {'name': 'Charlie', 'age': 35,},
    {'name': 'David', 'age': 40,}]

df = pd.DataFrame(data)
df['fruit_love'] = fruits

print(df)
```
