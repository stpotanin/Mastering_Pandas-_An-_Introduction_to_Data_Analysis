У вас есть DataFrame `orders`, содержащий информацию о заказах интернет-магазина. В нем имеются столбцы с информацией о номере заказа, дате заказа, имени покупателя, товарах, которые были заказаны, цене товаров и т.д. Однако вы обнаружили, что в отчете вам необходимо, чтобы столбец с именем покупателя был расположен первым, а все остальные столбцы следовали за ним. Таким образом, необходимо изменить порядок столбцов в DataFrame `orders` так, чтобы столбец с именем покупателя стоял первым, а все остальные столбцы следовали за ним.

Распечатайте обновленный DataFrame.

Типы данных в колонках не должны измениться.

```python
import pandas as pd
orders = pd.DataFrame({'order_id': [1, 2, 3, 4, 5],
        'order_date': ['2021-01-01', '2021-01-02', '2021-01-03', '2021-01-04', '2021-01-05'],
        'customer_name': ['John Doe', 'Jane Doe', 'Alex Smith', 'Sarah Jones', 'Bob Hope'],
        'product': ['Apple iPhone 12 Pro Max', 'Samsung Galaxy S21+', 'Sony Playstation 5 Console', 'Microsoft Xbox One', 'Apple Watch'],
        'price': [1100, 800, 500, 350, 400]})

orders = orders[['customer_name', 'order_id', 'order_date', 'product', 'price']]
print(orders)
```
