Дана таблица данных `dataframe_name`, содержащая информацию о продажах для нескольких продавцов за годы 2020 и 2021, включая идентификаторы клиентов и их имена. Необходимо выбрать только столбцы 'customer_id', 'customer_name' и 'sales_2021' из этой таблицы и на их основе создать новый DataFrame и вывести его на печать.

```python
import pandas as pd

dataframe_name = pd.DataFrame({'customer_id': [1, 2, 3, 4],
                               'customer_name': ['Alice', 'Bob', 'Charlie', 'David'],
                               'sales_2020': [1000, 1500, 1300, 2500],
                               'sales_2021': [1200, 1800, 1700, 2800]})

selection = dataframe_name[['customer_id', 'customer_name', 'sales_2021']]
print(selection)
```
