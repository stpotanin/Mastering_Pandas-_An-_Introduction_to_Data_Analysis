На вход в первой строке поступает целое число n, далее следуют n строк, в которых содержатся данные о производстве в формате `"Товар, Цена, Дата"`. Необходимо сформировать DataFrame, в котором каждый столбец соответствует одному из этих полей.

Выведите DataFrame на экран.

```python
import pandas as pd

data = [input().split(', ') for _ in range(int(input()))]

df = pd.DataFrame(data, columns=['Товар', 'Цена', 'Дата'])

print(df)

'''
2
Хлеб, 100, 2022-05-01
Молоко, 50, 2022-05-02
'''
print()
```
