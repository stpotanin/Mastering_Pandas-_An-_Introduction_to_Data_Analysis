Дана таблица данных `dataframe_name`, содержащая информацию о продажах для нескольких продавцов за годы 2020 и 2021. Необходимо изменить данные в столбце 'sales_2021' на новые значения, которые поступают во входной строке. Новые данные поступают в виде чисел, разделенных пробелом.

Выведите обновленный DataFrame на печать.

Тип данных в обновляемой колонке должен остаться `int`.

```python
import pandas as pd
dataframe_name = pd.DataFrame({'customer_id': [1, 2, 3, 4],
                               'customer_name': ['Alice', 'Bob', 'Charlie', 'David'],
                               'sales_2020': [1000, 1500, 1300, 2500],
                               'sales_2021': [1200, 1800, 1700, 2800]})

dataframe_name['sales_2021'] = [int(x) for x in input().split()]
print(dataframe_name)

# 2000 2500 3000 3500
```
