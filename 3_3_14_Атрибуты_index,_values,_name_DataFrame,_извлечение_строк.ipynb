Допустим, у нас есть список словарей `sales`, где каждый словарь представляет собой запись о продажах в определенном магазине за определенный период времени. Нам нужно создать DataFrame из этого списка, где каждой записи будет соответствовать строка, а в качестве индекса мы будем использовать даты продаж, которые мы возьмем из определенной колонки в списке словарей. При этом мы хотим, чтобы наш индекс имел имя `"date"`.

В результате мы получим DataFrame, в котором каждая строка будет соответствовать записи о продажах в определенный день, а индексом будет дата продаж.

Выведите итоговый DataFrame на экран.

```python
import pandas as pd

sales = [{'date': '2023-05-01', 'store': 'Магазин №1', 'sales': 1000},    {'date': '2023-05-01', 'store': 'Магазин №2', 'sales': 1500},    {'date': '2023-05-02', 'store': 'Магазин №1', 'sales': 2000},    {'date': '2023-05-02', 'store': 'Магазин №2', 'sales': 1800},    {'date': '2023-05-03', 'store': 'Магазин №1', 'sales': 3000},    {'date': '2023-05-03', 'store': 'Магазин №2', 'sales': 2700},    {'date': '2023-05-04', 'store': 'Магазин №1', 'sales': 4000},    {'date': '2023-05-04', 'store': 'Магазин №2', 'sales': 3200},]

df = pd.DataFrame(sales)
df.set_index('date', inplace=True)
print(df)
```
