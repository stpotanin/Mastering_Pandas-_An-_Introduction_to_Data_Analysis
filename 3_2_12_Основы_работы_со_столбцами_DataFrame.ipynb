Дана таблица данных с именем `dataframe_name`, содержащая информацию о продажах для нескольких продавцов за два года - 2019 и 2020. Необходимо переименовать столбцы 'sales_2019' и 'sales_2020' на 'previous_sales' и 'current_sales' соответственно в этой таблице.

Распечатайте обновленный DataFrame.

```python
import pandas as pd

dataframe_name = pd.DataFrame({'name': ['Alice', 'Bob', 'Charlie', 'David'],
                               'sales_2019': [1000, 1500, 1300, 2500],
                               'sales_2020': [1200, 1800, 1700, 2800]})

dataframe_name.rename(columns={'sales_2019': 'previous_sales', 'sales_2020': 'current_sales'}, inplace=True)

print(dataframe_name)
```
