Существует список списков data с данными и список `cols` с колонками, создайте на его основе `DataFrame` и выведите его на печать.

```python
import pandas as pd

date, duration = zip(*(x.split(':') for x in input().split()))

df = pd.DataFrame({'date': date, 'duration': duration})

print(df)

# 2015-08-03:70 2007-11-13:114 2018-09-08:77 2017-04-06:116 2001-01-09:117 2008-04-18:23 2012-01-12:28 2009-11-05:65 2004-06-25:52
```
