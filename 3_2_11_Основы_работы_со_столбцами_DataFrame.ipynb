Дана таблица данных с именем `dataframe_name`, содержащая информацию о продажах для нескольких продавцов за два года - 2020 и 2021. Необходимо изменить имя столбца 'sales_2020' на 'sales_2021' в этой таблице.

```python
import pandas as pd

dataframe_name = pd.DataFrame({'name': ['Alice', 'Bob', 'Charlie', 'David'],
                               'sales_2020': [1200, 1800, 1700, 2800],
                               'sales_2021': [1500, 2000, 1900, 3200]})

dataframe_name.rename(columns={'sales_2020': 'sales_2021'}, inplace=True)

print(dataframe_name)
```
