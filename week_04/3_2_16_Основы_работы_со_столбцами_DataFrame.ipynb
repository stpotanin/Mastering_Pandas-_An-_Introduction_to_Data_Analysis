Дана таблица `data` с информацией о пяти людях разного возраста и пола, работающих в разных странах мира и получающих заработную плату в двух разных годах.

John из США - молодой специалист, начинающий свою карьеру и увеличивающий свою зарплату каждый год. Maggie из Канады - успешная женщина в возрасте, которая уже достигла пика своей карьеры. Mark из Великобритании - опытный работник, который получает высокую зарплату благодаря своим навыкам и опыту. Jessica из Германии - молодая женщина, начинающая свою карьеру и надеющаяся получить повышение в ближайшее время. Will из Франции - специалист среднего уровня, который работает в компании уже несколько лет и получает стабильную зарплату. Необходимо:

1. Cоздать DataFrame `df` на ее основе.
2. Вывести первые 3 строки таблицы `df`.
3. Удалить столбец с информацией о стране.
4. Создать новый столбец `total_salary`, который будет содержать общую зарплату за два года.
5. Изменить название столбца с информацией о поле на `Sex`.
6. Вывести итоговый DataFrame на экран.

```python
import pandas as pd

data = {'Name': ['John', 'Maggie', 'Mark', 'Jessica', 'Will'],
        'Age': [23, 34, 45, 28, 31],
        'Gender': ['M', 'F', 'M', 'F', 'M'],
        'Country': ['USA', 'Canada', 'UK', 'Germany', 'France'],
        'Salary_2021': [45000, 52000, 60000, 41000, 48000],
        'Salary_2022': [47000, 55000, 61000, 42000, 50000]}

df = pd.DataFrame(data)
print(df.head(3))

df.drop('Country', axis=1, inplace=True)
df['total_salary'] = df.Salary_2021 + df.Salary_2022
df.rename(columns={'Gender': 'Sex'}, inplace=True)
print(df)
```
