На вход в первой строке поступает целое число n, далее следуют n строк, в которых содержатся данные о курсе валют в формате `"Дата, Валюта1/Валюта2, Курс"`. Необходимо сформировать DataFrame, в котором каждый столбец соответствует одному из этих полей.

Выведите DataFrame на экран.

```python
import pandas as pd

data = [input().split(', ') for _ in range(int(input()))]

df = pd.DataFrame(data, columns=['Дата', 'Валюта1/Валюта2', 'Курс'])

print(df)

'''
2
2023-05-05, USD/EUR, 0.824
2023-05-05, USD/JPY, 110.123
'''
print()
```
