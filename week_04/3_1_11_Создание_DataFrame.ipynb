Вы работаете в отделе маркетинга крупной компании, которая занимается продажей косметических средств. Вам необходимо проанализировать данные о продажах за последний месяц и выявить наиболее популярные продукты. Вам интересно узнать, сколько продуктов "Дневной крем для лица" было продано за этот период. Для этого вам нужно выбрать соответствующую строку в DataFrame с информацией о продажах, используя метод loc. Результат вывести на экран.

```python
import pandas as pd

data = [input().split(', ') for _ in range(3)]

df = pd.DataFrame(data, columns=['Имя Фамилия', 'Телефон', 'Email'])

print(df)

'''
John Doe, +1 555-123-4567, john.doe@example.com
Jane Smith, +44 20 1234 5678, jane.smith@example.co.uk
Alex Johnson, +61 2 1234 5678, aj@example.com
'''
print()
```

```python
# @title Правильный вариант
import pandas as pd

data = []
for _ in range(3):
    row = input().split(', ')
    name_parts = row[0].split(' ')
    first_name = name_parts[0]
    last_name = name_parts[1]
    data.append([first_name, last_name, row[1], row[2]])

df = pd.DataFrame(data, columns=['Имя', 'Фамилия', 'Телефон', 'Email'])

print(df)
```
