Представьте, что вы аналитик в компании, которая занимается производством мебели. У вас есть таблица с данными о продажах мебели за последний год sales_data, которую вы используете для анализа популярных продуктов и рыночной доли компании. Вам нужно вывести на печать таблицу в виде массива NumPy, чтобы увидеть общее количество продаж за год. Эта информация будет использоваться для создания графиков и диаграмм для дальнейшего анализа.

```python
import pandas as pd

# Исходные данные о продажах мебели
sales_data = {
    'Product': ['Стул', 'Стол', 'Диван', 'Кровать', 'Шкаф'],
    'Q1': [100, 75, 50, 120, 60],
    'Q2': [90, 65, 70, 100, 80],
    'Q3': [110, 85, 60, 130, 75],
    'Q4': [95, 80, 75, 110, 70]
}
# Создаем DataFrame и назначаем индексы
df = pd.DataFrame(sales_data)
print(df.to_numpy())
```
