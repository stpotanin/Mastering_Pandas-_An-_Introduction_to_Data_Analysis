У вас есть список кортежей, содержащих информацию о продажах на веб-сайте: дата продажи, название продукта, количество продуктов и общая сумма продажи. Создайте DataFrame на основе этого списка и назначьте индексами для строк даты, колонки назовите `'product_name'`, `'amount'`, `'sales_amount'`. Распечатайте DataFrame.

Используя метод `loc`, извлеките информацию только о продажах за 27 апреля 2023 года. Результат выведите на экран.

Примечание

> Обратите внимание, у индекса в итоговом DataFrame нет названия. Чтобы такого добиться, при создании DataFrame для колонки, которая далее будет использоваться в качестве индекса, задайте значение `None`. Потом установите эту колонку в качестве индекса с помощью `set_index`.



```python
import pandas as pd

sales_data = [('2023-04-25', 'Product A', 10, 500), ('2023-04-26', 'Product B', 5, 200), ('2023-04-27', 'Product C', 15, 1000), ('2023-04-28', 'Product D', 7, 350), ('2023-04-29', 'Product E', 20, 1200), ('2023-04-30', 'Product F', 12, 600), ('2023-05-01', 'Product G', 3, 150), ('2023-05-02', 'Product H', 8, 400), ('2023-05-03', 'Product I', 9, 450), ('2023-05-04', 'Product J', 6, 300), ('2023-05-05', 'Product K', 18, 900), ('2023-05-06', 'Product L', 11, 550)]

df = pd.DataFrame(sales_data, columns=[None, 'product_name', 'amount', 'sales_amount'])
df = df.set_index(df.columns[0])

print(df)
print(df.loc['2023-04-27'])
```
