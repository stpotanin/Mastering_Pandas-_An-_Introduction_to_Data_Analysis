На вход даны 5 строк входных данных, содержащих информацию о продажах для нескольких продавцов за два года - 2019 и 2020 в формате `name age sex sales_2019 sales_2020 region`. Необходимо создать таблицу DataFrame на основе этих данных c соответствующими названиями колонок и добавить новый столбец с именем `'total_sales'`, который будет являться суммой значений в столбцах `'sales_2019'` и `'sales_2020'`.

Выведите итоговый DataFrame.

Примечание 1. Тип данных в колонках `age`, `sales_2019`, `sales_2020` , `total_sales` должен быть `int` иначе система не пропустит решение.

```python
import pandas as pd

data = [input().split() for _ in range(5)]

columns = ['name', 'age', 'sex', 'sales_2019', 'sales_2020', 'region']
df = pd.DataFrame(data, columns=columns)
df[['age', 'sales_2019', 'sales_2020']] = df[['age', 'sales_2019', 'sales_2020']].astype(int)
df['total_sales'] = df['sales_2019'] + df['sales_2020']
print(df)

# Alice 25 female 1000 1200 North
# Bob 30 male 2000 1800 South
# Charlie 35 male 1500 1700 East
# David 40 male 3000 2800 West
# Ella 45 female 2500 2300 North
```
