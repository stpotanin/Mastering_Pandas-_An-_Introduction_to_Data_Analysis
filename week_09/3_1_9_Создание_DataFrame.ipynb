Имеется список списков `data`, создайте на его основе DataFrame и выведите его на печать. Затем задайте названия колонок для этого DataFrame: `'Имя', 'Возраст', 'Пол'`. Распечатайте обновленный DataFrame.

```python
import pandas as pd

data = [["John", 25, "Male"],
    ["Mary", 32, "Female"],
    ["Bob", 47, "Male"],
    ["Alice", 19, "Female"]]

df = pd.DataFrame(data)
print(df)

df.columns = ['Имя', 'Возраст', 'Пол']
print(df)
```
