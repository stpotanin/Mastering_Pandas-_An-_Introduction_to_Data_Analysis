На вход поступают 3 строки, которая содержит данные о продажах в формате `"Название товара|Количество|Стоимость|Дата продажи"`. Необходимо сформировать DataFrame, в котором каждый столбец соответствует одному из этих полей.

Выведите DataFrame на печать.

```python
import pandas as pd

data = [input().split('|') for _ in range(3)]

df = pd.DataFrame(data, columns=['Название товара', 'Количество', 'Стоимость', 'Дата продажи'])

print(df)

'''
iPhone 13|3|3499|2022-04-15
MacBook Pro|2|7999|2022-04-16
iPad Air|5|1999|2022-04-18
'''
print()
```
