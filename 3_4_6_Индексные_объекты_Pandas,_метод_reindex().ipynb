Пусть имеется список списков `data`, содержащий информацию о продуктах. Каждый вложенный список содержит два элемента: название продукта и количество.

Требуется создать DataFrame из этого списка, где название продукта будет значением индекса, а количество будет значением в соответствующей ячейке. Название индекса должно быть 'фрукты'. Распечатайте, получившийся DataFrame.

```python
import pandas as pd

data = [['яблоки', 10], ['апельсины', 15], ['бананы', 20]]

df = pd.DataFrame(data, columns=['фрукты', 'количество'])
df.set_index('фрукты', inplace=True)

print(df)
```
